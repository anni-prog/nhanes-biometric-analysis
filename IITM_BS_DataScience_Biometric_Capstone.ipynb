## Loading the NHANES Dataset

We load the male and female datasets into NumPy matrices.

## Statistical Summary of Weights

We compute measures of location (mean, median), dispersion (standard deviation), and shape (skewness, kurtosis).  
This helps us understand the distribution characteristics of male and female weights.

```python
def describe_weights(data, label):
    print(f"\n{label} statistics:")
    print(f"  Mean:           {np.mean(data):.4f}")
    print(f"  Median:         {np.median(data):.4f}")
    print(f"  Std Dev:        {np.std(data):.4f}")
    # Calculate skewness manually
    centered = data - np.mean(data)
    skewness = np.mean(centered**3) / (np.std(data)**3)
    print(f"  Skewness:       {skewness:.4f}")
    # Calculate kurtosis manually
    kurtosis = np.mean(centered**4) / (np.std(data)**4) - 3
    print(f"  Kurtosis:       {kurtosis:.4f}")

print("=" * 60)
print("DETAILED WEIGHT DISTRIBUTION ANALYSIS")
print("=" * 60)
describe_weights(female[:,0], "Female")
describe_weights(male[:,0], "Male")
```

## Scatterplot Matrix and Correlations

We visualize relationships between standardized variables (weight, height, waist, hip, BMI).  
We also compute Pearson and Spearman correlation coefficients to quantify linear and rank-based relationships.

```python
print("=" * 60)
print("CORRELATION ANALYSIS (Pearson Correlations)")
print("=" * 60)

# Column names
columns = ["Weight", "Height", "ArmLen", "LegLen", "ArmCirc", "HipCirc", "WaistCirc", "BMI"]

# Calculate Pearson correlation coefficients manually
def pearson_correlation(x, y):
    return np.corrcoef(x, y)[0, 1]

# Focus on key variables for correlation
key_cols = [0, 1, 6, 5, 7]  # Weight, Height, WaistCirc, HipCirc, BMI
key_names = ["Weight", "Height", "WaistCirc", "HipCirc", "BMI"]

print("\nKey Variable Correlations (Standardized):")
for i, col_i in enumerate(key_cols):
    for j, col_j in enumerate(key_cols):
        if i < j:  # Only upper triangle
            corr = pearson_correlation(zfemale[:, col_i], zfemale[:, col_j])
            print(f"  {key_names[i]:10s} vs {key_names[j]:10s}: {corr:7.4f}")

print("\nBMI Correlations with body measurements:")
bmi_col = 7
for i, name in enumerate(["Weight", "Height", "HipCirc", "WaistCirc"]):
    corr = pearson_correlation(zfemale[:, i], zfemale[:, bmi_col])
    print(f"  BMI vs {name:10s}: {corr:7.4f}")
```
